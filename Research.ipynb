```python
import matplotlib.pyplot as plt
import numpy as np
import math
```

```python
def calculate_projectile(v0, angle_deg, g=9.81):
  v0x = v0 * np.cos(np.radians(angle_deg))
  v0y = v0 * np.sin(np.radians(angle_deg))
  t = 2*v0y/g
  time_vals = np.linspace(0, t, 100)
  x = v0x * time_vals
  y = v0y * time_vals - 0.5*g*time_vals**2
  return (x, y, t)
```

```python
def calculate_with_air_resistence(v0, angle_deg, b, dt, g=9.81):
  x = [0.0]
  y = [0.0]
  rads = np.radians(angle_deg)

  vx = [v0*np.cos(rads)]
  vy = [v0*np.sin(rads)]
  v = v0

  time = 0
  while y[-1] >= 0:
    ax = -b * v * vx[-1]
    ay = -g -b * v * vy[-1]

    vx_new = vx[-1] + ax*dt
    vy_new = vy[-1] + ay*dt


    vx.append(vx_new)
    vy.append(vy_new)

    v = math.sqrt(vx_new**2 + vy_new**2)

    x.append(x[-1] + vx[-1]*dt)
    y.append(y[-1] + vy[-1]*dt)
    time += dt

  return (x, y, abs(np.degress(np.arctan(vy[-1]/vx[-1]))))

```

```python
launch_angles = [15,30,45,60,75]
v0 = 25

for angle in launch_angles:
  x, y = calculate_projectile(v0, angle)
  plt.plot(x, y, label=f"{angle}°")

plt.ylim(bottom=0)
plt.xlim(left=0)
plt.xlabel("Horizontal Distance (m)")
plt.ylabel("Vertical Distance (m)")
plt.title("Projectiles Trajectories at Various Launch Angles")
plt.legend()
plt.grid(True)

plt.show()
```

```python
vi = 25
deg = 30
b = 0.02
dt = 0.001

x, y, time = calculate_projectile(vi, deg)
with_resistx, with_resisty, time_resist, impact_angle = calculate_with_air_resistence(vi, deg, b, dt)

max_y = max(y)
with_resist_max_y = max(with_resisty)

max_x = max(x)
with_resist_max_x = max(with_resistx)

impace_ang = np.arctan()

print(f"""
==================================================
              FLIGHT METRICS SUMMARY
==================================================
Metric               Ideal (No Drag)   With Drag
--------------------------------------------------
Max Height (m):      {max_y}           {with_resist_max_y}
Total Range (m):     {max_x}           {with_resist_max_x}
Flight Time (s):     {time}            {time_resist}
Impact Angle (°):    {deg}             {impace_ang}
==================================================
""")
```
